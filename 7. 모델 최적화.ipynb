```python
import pandas as pd

redwine = pd.read_csv('winequality-red.csv', sep=';')
X, y = redwine.iloc[:, :-1], redwine.iloc[:, -1]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, 
                                                    stratify=y, # stratified sampling
                                                    test_size=0.25, # 75:25 split by default
                                                    random_state=1) 
```

# 기본 모델

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Input

model = Sequential()
model.add(Input(shape=(X_train.shape[1],)))
model.add(Dense(100, activation='sigmoid'))
model.add(Dense(200, activation='relu'))
model.add(Dense(50, activation='tanh'))
model.add(Dense(10, activation='softmax'))
model.summary()
```

```python
model.compile(loss='sparse_categorical_crossentropy', 
              optimizer='sgd', metrics=['accuracy'])
```

```python
from tensorflow.keras.callbacks import ModelCheckpoint, TensorBoard

checkpoint = ModelCheckpoint('best_model.keras', monitor='val_accuracy', save_best_only=True)
tensorboard = TensorBoard(log_dir='logs', histogram_freq=1, embeddings_freq=1)
```

```python
history = model.fit(X_train, y_train, epochs=300, batch_size=200, 
                    validation_data=(X_test, y_test),
                    callbacks=[checkpoint, tensorboard]
                    )
# train, val, test
# 학습, 검증, 테스트
# train: 학습용 데이터셋
# val: 검증용 데이터셋, 매 이폭마다 학습한 데이터의 성능을 평가하는 용도
# test: 테스트용 데이터셋, 학습이 완료된 후 최종적으로 모델의 성능을 평가하는 용도
```

```python
model.evaluate(X_test, y_test) # last model
```

```python
from tensorflow.keras.models import load_model
best_model = load_model('best_model.keras')

loss, accuracy = best_model.evaluate(X_test, y_test)
print(f'loss: {loss}, accuracy: {accuracy}')
```
